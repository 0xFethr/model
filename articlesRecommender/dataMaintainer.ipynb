```python
import pandas as pd
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from datetime import datetime
```

```python
data = pd.read_csv('Dataset/News_Category_Dataset_v3.csv', index_col=False)
```

```python
input = {"headline": "Over 4 Million Americans Roll Up Sleeves For Omicron-Targeted COVID Boosters", "category": "U.S. NEWS", "authors": "Carla K. Johnson, AP", "id":"9809", "date": "2022-09-23"}
```

```python
headline = input["headline"]
```

```python
stop_words = set(stopwords.words('english'))
string = ""
for word in headline.split():
    word = ("".join(e for e in word if e.isalnum()))
    word = word.lower()
    if not word in stop_words:
        string += word + " " 
headline = string.strip()
```

```python
lemmatizer = WordNetLemmatizer()
string = ""
for w in word_tokenize(headline):
    string += lemmatizer.lemmatize(w,pos = "v") + " "
headline = string.strip()
```

```python
input["headline"] = headline
input["date"] = datetime.strptime(input["date"], '%Y-%m-%d')
input["date"] = input["date"].strftime("%a") + "_" + input["date"].strftime("%b")
```

```python
data.loc[len(data.index)] = [input["headline"], input["category"], input["authors"], input["id"], input["date"]] 
```

```python
data.to_csv('Dataset/News_Category_Dataset_v3.csv', index = False)
```
